```python
from insightface.app import FaceAnalysis
import cv2
```

```python
# Initialize SCRFD
app = FaceAnalysis(
    name="buffalo_l",
    providers=["CPUExecutionProvider"]
)

app.prepare(ctx_id=0, det_size=(640, 640))
```

```python
# Load image
# image = cv2.imread("../Datasets/group_photo.jpg")
image = cv2.imread("../datasets/images/group_photo.jpg")
```

```python
# Detect faces
faces = app.get(image)

for face in faces:
    # Bounding Box
    bbox = face.bbox.astype(int)
    x1, y1, x2, y2 = bbox

    # Draw rectangle
    cv2.rectangle(
        image,
        (x1, y1),
        (x2, y2),
        (0, 255, 0),
        2
    )

    # Confidence score
    score = face.det_score
    cv2.putText(
        image,
        f"{score:.2f}",
        (x1, y1 - 10),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.6,
        (0, 255, 0),
        2
    )

    # Draw landmarks
    for point in face.kps.astype(int):
        cv2.circle(
            image,
            tuple(point),
            3,
            (0, 0, 255),
            -1
        )
```

```python
# Save output
cv2.imwrite("../datasets/results/SCRFD_detector_output.jpg", image)
```

```python
# Save Detection Results
for i, face in enumerate(faces):
    bbox = face.bbox.astype(int)

    x1, y1, x2, y2 = bbox

    face_crop = image[y1:y2, x1:x2]

    cv2.imwrite(
        f"../datasets/extracted_faces_SCRFD/face_{i}.jpg",
        face_crop
    )
```

#### **Better approach: Save aligned faces**

```python
import cv2
from insightface.app import FaceAnalysis
from insightface.utils import face_align

app = FaceAnalysis(
    name="buffalo_l",
    providers=["CPUExecutionProvider"]
)

app.prepare(ctx_id=0, det_size=(640, 640))

image = cv2.imread("../Datasets/group_photo.jpg")

faces = app.get(image)

for i, face in enumerate(faces):

    aligned_face = face_align.norm_crop(
        image,
        landmark=face.kps,
        image_size=112
    )

    cv2.imwrite(
        f"../datasets/extracted_faces_SCRFD/face_{i}.jpg",
        aligned_face
    )
```
