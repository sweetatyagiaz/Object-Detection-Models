This is on DeepFace dataset

```python
from adaface import AdaFace
```

```python
MODEL_PATH="../models/adaface_ir50.onnx"
DATABASE_DIR = "../datasets/dataset_deepface"
INDEX_PATH = "../datasets/fiss_indexes/face_index_adaface.bin"
MAPPING_PATH="../datasets/fiss_indexes/mapping_adaface.json"
```

```python
adaface = AdaFace(
    model_path=MODEL_PATH,
    database_dir=DATABASE_DIR,
    index_path=INDEX_PATH,
    mapping_path=MAPPING_PATH
)
```

```python
# Create index once
adaface.build_index()
```

```python
# Search
image_path = "../datasets/extracted_faces/vimala_face_1.jpg"

results = adaface.search_face(
    image_path,
    top_k=5,
    threshold=0.5
)
```

```python
for result in results:
    print(result)
```

```python
import os
from pathlib import Path

# Define your folder path
FOLDER_PATH = "../datasets/extracted_faces"

# Method 1: Using pathlib (Modern & Clean)
# This handles both lowercase '.jpg' and uppercase '.JPG' or '.jpeg'
jpg_files = []
for ext in ("*.jpg", "*.jpeg", "*.JPG", "*.JPEG"):
    jpg_files.extend(Path(FOLDER_PATH).rglob(ext))

print(f"Found {len(jpg_files)} JPG files:\n")
for file_path in jpg_files:
    print(f"Name: {file_path.name}")
    # print(f"Path: {file_path.resolve()}\n")  # .resolve() gives the absolute path

    results = adaface.search_face(file_path, top_k=5, threshold=0.5)
    for result in results:
        print(result)

    print('-'*25)
    print()

```
