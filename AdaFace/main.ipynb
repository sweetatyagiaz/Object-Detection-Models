```python
from adaface import AdaFace
```

```python
MODEL_PATH="../models/adaface_ir50.onnx"
DATABASE_DIR = "../datasets/extracted_faces_datasets"
INDEX_PATH = "../datasets/fiss_indexes/face_index_adaface.bin"
MAPPING_PATH="../datasets/fiss_indexes/mapping_adaface.json"
```

```python
adaface = AdaFace(
    model_path=MODEL_PATH,
    database_dir=DATABASE_DIR,
    index_path=INDEX_PATH,
    mapping_path=MAPPING_PATH
)
```

```python
# Create index once
adaface.build_index()
```

```python
# Search
image_path = "../datasets/extracted_faces_SCRFD/face_1.jpg"

results = adaface.search_face(
    image_path,
    top_k=5,
    threshold=0.5
)
```

```python
for result in results:
    print(result)
```
