```python
input_folder = "../raw_data/images"
output_folder = "../raw_data/extracted_faces"
faces_dir = output_folder
embeddings_dir = "../raw_data/embeddings"
faiss_emb_file = '../Datasets/faces_embedding.index'
```

#### **1. Extract faces**

```python
extract_faces(images_folder=input_folder, output_folder=output_folder)
```

#### **2. Create Face Embeddings**

```python
# Create Embedding of all images

embedding = create_embeddings(save_emb=True, faces_dir=faces_dir, embeddings_dir=embeddings_dir)
```

#### **3. Comapre Faces**

```python
# Compare Faces
face_emb1 = "../raw_data/embeddings/IMG_5315_face_1.npy"
face_emb2 = "../raw_data/embeddings/IMG_5315_face_1.npy"

compare_faces(face_emb1=face_emb1, face_emb2=face_emb2)
```

```python
compare_faces(face_emb1=embedding, face_emb2=face_emb2)
```

#### **4. Create FAISS Index**

```python
# dir_path = Path("raw_data/embeddings")
dir_path = Path(embeddings_dir)

# create FAISS Index
faiss_index = create_faiss_index()

# Add Embeddings
emb_files = [f for f in dir_path.rglob("*.npy") if f.is_file()]

for file in emb_files:
    print(file)  # Prints the full path relative to dir_path
    
    add_embeddings(faiss_index=faiss_index, file_path=file)
    
print(faiss_index.ntotal)
```

#### **5. Search Face**

```python
# Search face using embeding
search_face(faiss_index=faiss_index, search_emb_path='../raw_data/embeddings/IMG_5302_face_1.npy', embeddings_dir=embeddings_dir)
```

```python
# Search face using raw image
search_face(faiss_index=faiss_index, search_img_path='../raw_data/extracted_faces/IMG_5302_face_1.jpg', 
            embeddings_dir=embeddings_dir)
```

#### **6. FAISS: Save/Load**

```python
# Save Faces Embeddings
faiss_load_save(option_save=True, faiss_index=faiss_index, file_path=faiss_emb_file)
```

```python
# Load Faces Embeddings
faiss_index = faiss_load_save(file_path=faiss_emb_file)
```

```python
# Search face using embeding: after loading Facess Index
search_face(faiss_index=faiss_index, search_emb_path='../raw_data/embeddings/IMG_5302_face_1.npy', embeddings_dir=embeddings_dir)
```
