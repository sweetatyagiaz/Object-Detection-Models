```python
import json
from ultralytics import YOLO

# 1. Load the model and run inference
model = YOLO("../Datasets/yolo26n.pt")
results = model("../Datasets/multi_face.jpg", classes=[0])  # Filter for 'person'

# 2. Extract detection data
detections = []
for result in results:
    for box in result.boxes:
        # Convert coordinates to list of integers or floats
        bbox = [int(coord) for coord in box.xyxy[0].tolist()]
        confidence = float(box.conf[0].item())
        class_name = model.names[int(box.cls[0].item())]
        
        # Structure the detection dict
        detections.append({
            "class": class_name,
            "confidence": round(confidence, 2),
            "bbox": bbox
        })

# 3. Save to a JSON file
output_path = "../Datasets/multi_face.json"
with open(output_path, "w") as f:
    json.dump(detections, f, indent=2)

print(f"Results successfully saved to {output_path}")
```
