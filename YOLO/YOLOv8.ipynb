```python
from ultralytics import YOLO

# Load a pre-trained YOLOv8 nano model (downloads automatically)
model = YOLO("../Datasets/yolov8n.pt")

# Run inference on your image
results = model("../Datasets/multi_face.jpg", conf=0.5)

# Filter for 'person' class (class ID 0) and display/save results
for r in results:
    boxes = r.boxes
    for box in boxes:
        cls_id = int(box.cls[0])
        if model.names[cls_id] == "person":
            print(f"Person detected with confidence {box.conf[0]:.2f} at coordinates {box.xyxy[0]}")

# Save the annotated image with bounding boxes
results[0].save(filename="../Datasets/multi_face_output.jpg")
```
