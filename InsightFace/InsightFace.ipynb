##### **Extract Faces from a Single Image**

```python
import cv2
import os
from insightface.app import FaceAnalysis
```

```python
# input_image = "../datasets/images/multi_face_output.jpg"
input_image = "../datasets/images/group_photo_1.jpg"
output_dir = "../datasets/extracted_faces"
```

```python
# Initialize face detector
app = FaceAnalysis(
    providers=['CPUExecutionProvider']  # Use CUDAExecutionProvider if GPU available
)
app.prepare(ctx_id=0, det_size=(640, 640))

os.makedirs(output_dir, exist_ok=True)

# Read image
image = cv2.imread(input_image)

# Detect faces
faces = app.get(image)

print(f"Found {len(faces)} face(s)")

for i, face in enumerate(faces):
    x1, y1, x2, y2 = map(int, face.bbox)

    # Add small margin
    margin = 20
    x1 = max(0, x1 - margin)
    y1 = max(0, y1 - margin)
    x2 = min(image.shape[1], x2 + margin)
    y2 = min(image.shape[0], y2 + margin)

    face_crop = image[y1:y2, x1:x2]

    output_file = os.path.join(output_dir, f"face_{i+1}.jpg")
    cv2.imwrite(output_file, face_crop)

    print(f"Saved: {output_file}")
```

```python
import cv2
import os
from pathlib import Path
from insightface.app import FaceAnalysis

app = FaceAnalysis(providers=['CPUExecutionProvider'])
app.prepare(ctx_id=0, det_size=(640, 640))

# input_folder = "raw_data/images"
input_folder = "../datasets/images"
output_folder = "../datasets/extracted_faces"

os.makedirs(output_folder, exist_ok=True)

for image_file in Path(input_folder).glob("*.*"):
    if image_file.suffix.lower() not in [".jpg", ".jpeg", ".png", ".bmp", ".webp"]:
        continue

    image = cv2.imread(str(image_file))

    if image is None:
        continue

    faces = app.get(image)

    for i, face in enumerate(faces):
        x1, y1, x2, y2 = map(int, face.bbox)

        face_crop = image[y1:y2, x1:x2]

        output_name = (
            f"{image_file.stem}_face_{i+1}.jpg"
        )

        cv2.imwrite(
            os.path.join(output_folder, output_name),
            face_crop
        )

    print(f"{image_file.name}: {len(faces)} face(s)")
```
