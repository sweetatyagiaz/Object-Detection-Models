```python
import pandas as pd
```

```python
# Read a CSV file from the same directory
df = pd.read_csv('face_tracking_log.csv')

# View the first 5 rows
print(df.head())
```

#### **1. Person Attendance Summary**

```python
attendance = df.groupby(
    ["Person_ID", "Person_Name"]
).agg(
    First_Seen=("Timestamp", "min"),
    Last_Seen=("Timestamp", "max"),
    Total_Events=("Timestamp", "count"),
    Avg_Score=("Score", "mean"),
    Max_Score=("Score", "max")
).reset_index()

attendance
```

##### **2. Entry / Exit Analysis**

```python
entries = df[df["Status"] == "ENTER"]
exits = df[df["Status"] == "EXIT"]
```

```python
exits
```

#### **3. Dwell Time Analysis**

```python
# 1. Convert Timestamp strings to datetime objects
df["Timestamp"] = pd.to_datetime(df["Timestamp"])

# 2. Group by Track_ID, grab the person's name, and compute the dwell time
summary_df = df.groupby("Track_ID").agg(
    Person=("Person_Name", "first"),
    Dwell_Time=("Timestamp", lambda x: f"{int((x.max() - x.min()).total_seconds())} sec")
).reset_index()

# 3. Rename headers to match your desired format
summary_df.columns = ["Track", "Person", "Dwell Time"]

# 4. Print as a markdown table (Requires: pip install tabulate)
print(summary_df.to_markdown(index=False))
```

#### **4. Recognition Quality**

```python
score_analysis = df.groupby(
    ["Person_ID", "Person_Name"]
)["Score"].agg(
    ["mean", "min", "max", "std"]
)

score_analysis
```

#### **5. Camera Utilization**

```python
camera_stats = df.groupby("Camera_ID").agg(
    Events=("Timestamp", "count"),
    Persons=("Person_ID", "nunique"),
    Tracks=("Track_ID", "nunique")
)

camera_stats
```
