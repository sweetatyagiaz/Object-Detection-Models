#### **Extract Faces**

```python
def extract_faces(images_folder=False, output_folder=False):
    for image_file in Path(images_folder).glob("*"):
        if image_file.suffix.lower() not in [".jpg", ".jpeg", ".png", ".webp"]:
            continue

        try:
            faces = RetinaFace.extract_faces(
                img_path=str(image_file),
                align=True
            )

            for idx, face in enumerate(faces):
                face = cv2.cvtColor(face, cv2.COLOR_RGB2BGR)

                filename = f"{image_file.stem}_face_{idx+1}.jpg"
                cv2.imwrite(
                    os.path.join(output_folder, filename),
                    face
                )

            print(f"{image_file.name}: {len(faces)} faces")

        except Exception as e:
            print(f"Error: {image_file.name} -> {e}")

    return False
```

```python
# Extract faces
extract_faces(images_folder=input_folder, output_folder=output_folder)
```

#### **Face size**

```python
from retinaface import RetinaFace
import cv2


def calculate_face_size(file_path=False):
    # image = cv2.imread("Datasets/group_photo.jpg")

    if not file_path:
        return False
    
    image = cv2.imread(file_path)

    faces = RetinaFace.detect_faces(image)

    print(f"File: {file_path}")
    
    for face_id, face_data in faces.items():
        x1, y1, x2, y2 = face_data["facial_area"]

        width = x2 - x1
        height = y2 - y1

        print(f"Face Size: {width} x {height}")
```

#### **Compare Two Faces**

ArcFace embeddings are designed for cosine-similarity matching in recognition, search, and watchlist workflows.

```python
def compare_faces(face_emb1=False, face_emb2=False):
    if isinstance(face_emb1, str):
        face_emb1 = np.load(face_emb1)
    if isinstance(face_emb2, str):
        face_emb2 = np.load(face_emb2)
    
    # Cosine Similarity
    similarity = np.dot(
        face_emb1 / np.linalg.norm(face_emb1),
        face_emb2 / np.linalg.norm(face_emb2)
    )

    print(f"Similarity: {similarity:.4f}")

    if similarity > 0.5:
        print("Same Person")
        return True
    else:
        print("Different Person")
        return False
```

#### **How to use FAISS**

Step 1: Generate ArcFace Embeddings

```python
import faiss
import numpy as np
from pathlib import Path
```

```python
# Step 2: Create a FAISS Index
def create_faiss_index(DIMENSION = 512):
    # Cosine similarity requires normalized vectors
    index = faiss.IndexFlatIP(DIMENSION)

    print(index.ntotal)

    return index
```

```python
def load_embedding(file_path=False):
    embedding = np.array([
            np.load(file_path),
        ], dtype=np.float32)
    
    # Normalize
    faiss.normalize_L2(embedding)

    return embedding

def add_embeddings(faiss_index=False, file_path=False):    
    embedding = load_embedding(file_path=file_path)

    # index.add(embedding)
    faiss_index.add(embedding)

    return True
```

```python
dir_path = Path("raw_data/embeddings")

# create FAISS Index
faiss_index = create_faiss_index()

# Add Embeddings
emb_files = [f for f in dir_path.rglob("*.npy") if f.is_file()]

for file in emb_files:
    print(file)  # Prints the full path relative to dir_path
    
    add_embeddings(faiss_index=faiss_index, file_path=file)
    
print(faiss_index.ntotal)
```

```python
def search_face(faiss_index=False, search_emb_path=False, search_img_path=False, threshold=0.6):
    if search_img_path:
        # query = create_embedding(file_path=search_img_path)
        # query = create_embedding_image(save_emb=False, faces_dir=search_img_path, embeddings_dir=embeddings_dir)
        query = create_embeddings(save_emb=False, faces_dir=search_img_path, embeddings_dir=embeddings_dir)
        # query = np.array(
        #             create_embeddings(save_emb=True, faces_dir=search_img_path, embeddings_dir=embeddings_dir),
        #             dtype=np.float32
        #             )
        query = query[np.newaxis, :]
        faiss.normalize_L2(query)
    else:
        query = np.array(
            load_embedding(file_path=search_emb_path),
            dtype=np.float32
            )
        # query = np.array(
        #     [search_emb_path],
        #     dtype=np.float32
        # )

        faiss.normalize_L2(query)

    distances, indices = faiss_index.search(
        query,
        5
    )

    score = float(distances[0][0])
    idx = int(indices[0][0])

    if score >= threshold:
        return indices, distances

    return None, distances
```
